```python
%sql

SELECT 'bronze_customer_info' AS table_name, COUNT(*) AS rows FROM retail_medallion_ws.default.bronze_customer_info
UNION ALL SELECT 'silver_customer', COUNT(*) FROM retail_medallion_ws.default.silver_customer
UNION ALL SELECT 'dim_customer', COUNT(*) FROM retail_medallion_ws.default.dim_customer
UNION ALL SELECT 'bronze_product_info', COUNT(*) FROM retail_medallion_ws.default.bronze_product_info
UNION ALL SELECT 'silver_product_info', COUNT(*) FROM retail_medallion_ws.default.silver_product_info
UNION ALL SELECT 'dim_product', COUNT(*) FROM retail_medallion_ws.default.dim_product
UNION ALL SELECT 'bronze_sales_details', COUNT(*) FROM retail_medallion_ws.default.bronze_sales_details
UNION ALL SELECT 'silver_sales', COUNT(*) FROM retail_medallion_ws.default.silver_sales
UNION ALL SELECT 'fact_sales', COUNT(*) FROM retail_medallion_ws.default.fact_sales
ORDER BY table_name;
```

```python
%sql

SELECT 'dim_customer' AS tbl, customer_sk, COUNT(*) c FROM retail_medallion_ws.default.dim_customer GROUP BY customer_sk HAVING COUNT(*) > 1
UNION ALL
SELECT 'dim_product', product_sk, COUNT(*) FROM retail_medallion_ws.default.dim_product GROUP BY product_sk HAVING COUNT(*) > 1
UNION ALL
SELECT 'dim_date', date_sk, COUNT(*) FROM retail_medallion_ws.default.dim_date GROUP BY date_sk HAVING COUNT(*) > 1;
```

```python
%sql

SELECT COUNT(*) AS fact_rows_missing_customer FROM retail_medallion_ws.default.fact_sales WHERE customer_sk IS NULL;
SELECT COUNT(*) AS fact_rows_missing_product FROM retail_medallion_ws.default.fact_sales WHERE product_sk IS NULL;
SELECT COUNT(*) AS fact_rows_missing_order_date FROM retail_medallion_ws.default.fact_sales WHERE order_date_sk IS NULL;
```

```python
%sql

SELECT COUNT(*) AS bad_unit_price FROM retail_medallion_ws.default.fact_sales WHERE unit_price <= 0;
SELECT COUNT(*) AS bad_quantity FROM retail_medallion_ws.default.fact_sales WHERE quantity <= 0;
SELECT COUNT(*) AS mismatched_sales_amount FROM retail_medallion_ws.default.fact_sales WHERE sales_amount <> quantity * unit_price;
```

```python
%sql

SELECT
    (SELECT COUNT(*) FROM retail_medallion_ws.default.fact_sales WHERE customer_sk IS NULL) AS orphan_customer_fk,
    (SELECT COUNT(*) FROM retail_medallion_ws.default.fact_sales WHERE product_sk IS NULL) AS orphan_product_fk,
    (SELECT COUNT(*) FROM retail_medallion_ws.default.fact_sales WHERE sales_amount <> quantity * unit_price) AS bad_sales_calc,
    (SELECT COUNT(*) FROM retail_medallion_ws.default.fact_sales) AS total_fact_rows;
```
