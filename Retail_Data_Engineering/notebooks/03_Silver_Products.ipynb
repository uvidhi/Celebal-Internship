```python
%sql

SELECT COUNT(*) AS total_rows
FROM retail_medallion_ws.default.bronze_product_info;
```

```python
%sql

SELECT
    COUNT(*) AS total_rows,
    SUM(CASE WHEN prd_id IS NULL THEN 1 ELSE 0 END) AS null_prd_id,
    SUM(CASE WHEN prd_key IS NULL THEN 1 ELSE 0 END) AS null_prd_key,
    SUM(CASE WHEN prd_nm IS NULL THEN 1 ELSE 0 END) AS null_prd_nm,
    SUM(CASE WHEN prd_cost IS NULL THEN 1 ELSE 0 END) AS null_prd_cost,
    SUM(CASE WHEN prd_line IS NULL THEN 1 ELSE 0 END) AS null_prd_line,
    SUM(CASE WHEN prd_start_dt IS NULL THEN 1 ELSE 0 END) AS null_prd_start_dt,
    SUM(CASE WHEN prd_end_dt IS NULL THEN 1 ELSE 0 END) AS null_prd_end_dt
FROM retail_medallion_ws.default.bronze_product_info;
```

```python
%sql

SELECT
    prd_id,
    COUNT(*) AS record_count
FROM retail_medallion_ws.default.bronze_product_info
GROUP BY prd_id
HAVING COUNT(*) > 1;
```

```python
%sql

SELECT COUNT(*) AS invalid_date_records
FROM retail_medallion_ws.default.bronze_product_info
WHERE prd_end_dt IS NOT NULL
  AND prd_end_dt < prd_start_dt;
```

```python
%sql

SELECT
    COUNT(*) AS invalid_date_records,
    MIN(prd_start_dt) AS earliest_start_date,
    MAX(prd_start_dt) AS latest_start_date,
    MIN(prd_end_dt) AS earliest_end_date,
    MAX(prd_end_dt) AS latest_end_date
FROM retail_medallion_ws.default.bronze_product_info
WHERE prd_end_dt IS NOT NULL
  AND prd_end_dt < prd_start_dt;
```

```python
%sql

CREATE OR REPLACE TABLE retail_medallion_ws.default.silver_product_info AS

SELECT
    prd_id,
    prd_key,
    prd_nm,
    prd_cost,
    prd_line,
    prd_start_dt,
    CASE
        WHEN prd_end_dt IS NOT NULL
             AND prd_end_dt < prd_start_dt
        THEN NULL
        ELSE prd_end_dt
    END AS prd_end_dt
FROM retail_medallion_ws.default.bronze_product_info
WHERE prd_id IS NOT NULL;
```

```python
%sql

SELECT *
FROM retail_medallion_ws.default.silver_product_info
LIMIT 10;
```

```python
%sql

SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT prd_id) AS unique_product_ids
FROM retail_medallion_ws.default.silver_product_info;
```

```python
%sql

SELECT COUNT(*) AS invalid_dates_remaining
FROM retail_medallion_ws.default.silver_product_info
WHERE prd_end_dt IS NOT NULL
  AND prd_end_dt < prd_start_dt;
```
