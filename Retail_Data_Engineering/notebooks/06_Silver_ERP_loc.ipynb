```python

loc_df_ = spark.table(
    "retail_medallion_ws.default.bronze_erp_location"
)

loc_df_.printSchema()
```

```python
display(loc_df_.limit(10))
```

```python
%sql

SELECT
    COUNT(*) AS total_rows,
    SUM(CASE WHEN CID IS NULL THEN 1 ELSE 0 END) AS null_CID,
    SUM(CASE WHEN CNTRY IS NULL THEN 1 ELSE 0 END) AS null_CNTRY
FROM retail_medallion_ws.default.bronze_erp_location;
```

```python
%sql

SELECT
    CID,
    COUNT(*) AS record_count
FROM retail_medallion_ws.default.bronze_erp_location
GROUP BY CID
HAVING COUNT(*) > 1;
```

```python
%sql

CREATE OR REPLACE TABLE retail_medallion_ws.default.silver_erp_location AS

SELECT
    CID,
    CNTRY
FROM retail_medallion_ws.default.bronze_erp_location
WHERE CNTRY IS NOT NULL;
```

```python
%sql

SELECT
    COUNT(*) AS total_rows,
    SUM(CASE WHEN CID IS NULL THEN 1 ELSE 0 END) AS null_CID,
    SUM(CASE WHEN CNTRY IS NULL THEN 1 ELSE 0 END) AS null_CNTRY
FROM retail_medallion_ws.default.silver_erp_location;
```
