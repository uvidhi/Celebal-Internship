```python

cat_df_ = spark.table(
    "retail_medallion_ws.default.bronze_erp_category"
)

cat_df_.printSchema()
```

```python
display(cat_df_.limit(10))
```

```python
%sql

SELECT
    COUNT(*) AS total_rows,
    SUM(CASE WHEN ID IS NULL THEN 1 ELSE 0 END) AS null_ID,
    SUM(CASE WHEN CAT IS NULL THEN 1 ELSE 0 END) AS null_CAT,
    SUM(CASE WHEN SUBCAT IS NULL THEN 1 ELSE 0 END) AS null_SUBCAT,
    SUM(CASE WHEN MAINTENANCE IS NULL THEN 1 ELSE 0 END) AS null_MAINTENANCE
FROM retail_medallion_ws.default.bronze_erp_category;
```

```python
%sql

SELECT
    ID,
    COUNT(*) AS record_count
FROM retail_medallion_ws.default.bronze_erp_category
GROUP BY ID
HAVING COUNT(*) > 1;
```

```python
%sql

CREATE OR REPLACE TABLE retail_medallion_ws.default.silver_erp_category AS

SELECT
    ID,
    CAT,
    SUBCAT,
    MAINTENANCE
FROM retail_medallion_ws.default.bronze_erp_category;
```

```python
%sql

SELECT COUNT(*) AS total_rows
FROM retail_medallion_ws.default.silver_erp_category;
```
