```python
%sql

SELECT
    COUNT(*) AS total_rows,
    SUM(CASE WHEN CID IS NULL THEN 1 ELSE 0 END) AS null_CID,
    SUM(CASE WHEN BDATE IS NULL THEN 1 ELSE 0 END) AS null_BDATE,
    SUM(CASE WHEN GEN IS NULL THEN 1 ELSE 0 END) AS null_GEN
FROM retail_medallion_ws.default.bronze_erp_customer;
```

```python
%sql

SELECT
    CID,
    COUNT(*) AS record_count
FROM retail_medallion_ws.default.bronze_erp_customer
GROUP BY CID
HAVING COUNT(*) > 1;
```

```python
%sql

CREATE OR REPLACE TABLE retail_medallion_ws.default.silver_erp_customer AS

SELECT
    CID,
    BDATE,
    GEN
FROM retail_medallion_ws.default.bronze_erp_customer
WHERE GEN IS NOT NULL;
```

```python
%sql

SELECT
    COUNT(*) AS total_rows,
    SUM(CASE WHEN CID IS NULL THEN 1 ELSE 0 END) AS null_CID,
    SUM(CASE WHEN BDATE IS NULL THEN 1 ELSE 0 END) AS null_BDATE,
    SUM(CASE WHEN GEN IS NULL THEN 1 ELSE 0 END) AS null_GEN
FROM retail_medallion_ws.default.silver_erp_customer;
```
