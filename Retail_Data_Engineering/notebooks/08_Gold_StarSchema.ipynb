```sql


CREATE OR REPLACE TABLE retail_medallion_ws.default.dim_customer AS
SELECT DISTINCT
    cst_id AS customer_id,
    cst_key AS customer_key,
    cst_firstname AS first_name,
    cst_lastname AS last_name,
    cst_gndr AS gender
FROM retail_medallion_ws.default.silver_customer;
```

```sql
-- Check customer data

SELECT *
FROM retail_medallion_ws.default.dim_customer
LIMIT 10;
```

```sql


CREATE OR REPLACE TABLE retail_medallion_ws.default.dim_product AS
SELECT DISTINCT
    prd_id AS product_id,
    prd_key AS product_key,
    prd_nm AS product_name,
    prd_cost AS product_cost,
    prd_line AS product_line
FROM retail_medallion_ws.default.silver_product_info;
```

```sql


SELECT *
FROM retail_medallion_ws.default.dim_product
LIMIT 10;
```

```sql


CREATE OR REPLACE TABLE retail_medallion_ws.default.fact_sales AS
SELECT
    sls_ord_num AS order_id,
    sls_prd_key AS product_key,
    sls_cust_id AS customer_id,
    sls_order_dt AS order_date,
    sls_quantity AS quantity,
    sls_price AS unit_price,
    sls_sales AS sales_amount
FROM retail_medallion_ws.default.silver_sales;
```

```sql


SELECT *
FROM retail_medallion_ws.default.fact_sales
LIMIT 10;
```

```sql


SHOW TABLES IN retail_medallion_ws.default;
```

```sql


SELECT 'dim_customer' AS table_name, COUNT(*) AS records
FROM retail_medallion_ws.default.dim_customer

UNION ALL

SELECT 'dim_product', COUNT(*)
FROM retail_medallion_ws.default.dim_product

UNION ALL

SELECT 'fact_sales', COUNT(*)
FROM retail_medallion_ws.default.fact_sales;
```
