```python
%sql

SELECT
    SUM(sales_amount) AS total_revenue
FROM retail_medallion_ws.default.fact_sales;
```

```python
%sql

SELECT
    COUNT(DISTINCT order_id) AS total_orders
FROM retail_medallion_ws.default.fact_sales;
```

```python
%sql

SELECT
    ROUND(SUM(sales_amount) / COUNT(DISTINCT order_id), 2) AS average_order_value
FROM retail_medallion_ws.default.fact_sales;
```

```python
%sql

SELECT
    p.product_name,
    SUM(f.sales_amount) AS total_sales
FROM retail_medallion_ws.default.fact_sales f
JOIN retail_medallion_ws.default.dim_product p
    ON f.product_key = SUBSTRING(p.product_key, 7, LEN(p.product_key))
GROUP BY p.product_name
ORDER BY total_sales DESC
LIMIT 10;
```

```python
%sql

SELECT
    c.first_name,
    c.last_name,
    SUM(f.sales_amount) AS total_spent
FROM retail_medallion_ws.default.fact_sales f
JOIN retail_medallion_ws.default.dim_customer c
    ON f.customer_id = c.customer_id
GROUP BY
    c.first_name,
    c.last_name
ORDER BY total_spent DESC
LIMIT 10;
```

```python
%sql

SELECT
    p.product_line,
    SUM(f.sales_amount) AS total_sales
FROM retail_medallion_ws.default.fact_sales f
JOIN retail_medallion_ws.default.dim_product p
    ON f.product_key = SUBSTRING(p.product_key, 7, LEN(p.product_key))
GROUP BY p.product_line
ORDER BY total_sales DESC;
```

```python
%sql

SELECT
    c.customer_id,
    c.first_name,
    c.last_name,
    SUM(f.sales_amount) AS total_spent,
    CASE
        WHEN SUM(f.sales_amount) >= 10000 THEN 'High Value'
        WHEN SUM(f.sales_amount) >= 5000 THEN 'Medium Value'
        ELSE 'Low Value'
    END AS customer_segment
FROM retail_medallion_ws.default.fact_sales f
JOIN retail_medallion_ws.default.dim_customer c
    ON f.customer_id = c.customer_id
GROUP BY
    c.customer_id,
    c.first_name,
    c.last_name
ORDER BY total_spent DESC;
```
