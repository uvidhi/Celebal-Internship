```python
invalid_emails = customers_df[
    ~customers_df["email"].str.contains(
        "@",
        na=False
    )
]

print("Invalid emails:", len(invalid_emails))

display(invalid_emails)
```

```python
def validate_emails(df):
    df = df.copy()

    email_pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"

    valid_email = df["email"].str.match(
        email_pattern,
        na=False
    )

    df.loc[~valid_email, "email"] = None

    return df
```

```python
customers_df = validate_emails(customers_df)
```

```python
print("Invalid emails:", customers_df["email"].isna().sum())
```

```python
def clean_products(df):
    df = df.copy()

    df["product_name"] = (
        df["product_name"]
        .astype(str)
        .str.strip()
        .str.title()
    )

    df["subcategory"] = (
        df["subcategory"]
        .astype(str)
        .str.strip()
        .str.title()
    )

    return df
```

```python
products_df = clean_products(products_df)
```

```python
display(products_df.head(10))
```

```python
orders_df["order_date"] = pd.to_datetime(
    orders_df["order_date"],
    dayfirst=True,
    errors="coerce"
)
```

```python
print(orders_df["order_date"].dtype)
```

```python
print("Orders before:", len(orders_df))

orders_df = orders_df.dropna(
    subset=["customer_id"]
)

print("Orders after:", len(orders_df))
```

```python
valid_order_ids = set(orders_df["order_id"])

invalid_items = order_items_df[
    ~order_items_df["order_id"].isin(valid_order_ids)
]

print("Invalid order items:", len(invalid_items))
```

```python
order_items_df = order_items_df[
    order_items_df["order_id"].isin(valid_order_ids)
].copy()
```
