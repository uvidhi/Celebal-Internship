```python
%sql
SELECT
    c.customer_id,
    c.customer_name,
    COUNT(DISTINCT o.order_id) AS total_orders
FROM workspace.default.customers c
LEFT JOIN workspace.default.orders o
    ON c.customer_id = o.customer_id
GROUP BY
    c.customer_id,
    c.customer_name
ORDER BY total_orders DESC;
```

```python
%sql
WITH customer_orders AS (
    SELECT
        c.customer_id,
        c.customer_name,
        COUNT(DISTINCT o.order_id) AS total_orders
    FROM workspace.default.customers c
    LEFT JOIN workspace.default.orders o
        ON c.customer_id = o.customer_id
    GROUP BY
        c.customer_id,
        c.customer_name
)

SELECT
    customer_id,
    customer_name,
    total_orders,

    CASE
        WHEN total_orders = 0 THEN 'No Purchase'
        WHEN total_orders = 1 THEN 'One-time'
        WHEN total_orders <= 4 THEN 'Occasional'
        ELSE 'Loyal'
    END AS purchase_segment

FROM customer_orders
ORDER BY total_orders DESC;
```

```python
%sql
SELECT
    c.customer_id,
    c.customer_name,

    ROUND(
        COALESCE(
            SUM(
                oi.quantity
                * oi.unit_price
                * (1 - oi.discount_percent / 100)
            ),
            0
        ),
        2
    ) AS total_spend

FROM workspace.default.customers c

LEFT JOIN workspace.default.orders o
    ON c.customer_id = o.customer_id

LEFT JOIN workspace.default.order_items oi
    ON o.order_id = oi.order_id

GROUP BY
    c.customer_id,
    c.customer_name

ORDER BY total_spend DESC;
```

```python
%sql
WITH customer_spending AS (
    SELECT
        c.customer_id,
        c.customer_name,

        COALESCE(
            SUM(
                oi.quantity
                * oi.unit_price
                * (1 - oi.discount_percent / 100)
            ),
            0
        ) AS total_spend

    FROM workspace.default.customers c

    LEFT JOIN workspace.default.orders o
        ON c.customer_id = o.customer_id

    LEFT JOIN workspace.default.order_items oi
        ON o.order_id = oi.order_id

    GROUP BY
        c.customer_id,
        c.customer_name
)

SELECT
    customer_id,
    customer_name,
    ROUND(total_spend, 2) AS total_spend,

    CASE
        WHEN total_spend = 0 THEN 'No Purchase'
        WHEN total_spend < 1000 THEN 'Low'
        WHEN total_spend <= 5000 THEN 'Medium'
        ELSE 'High'
    END AS spend_tier

FROM customer_spending
ORDER BY total_spend DESC;
```

```python
%sql
SELECT
    c.customer_id,
    c.customer_name,

    MAX(o.order_date) AS last_purchase_date,

    COUNT(DISTINCT o.order_id) AS frequency,

    ROUND(
        COALESCE(
            SUM(
                oi.quantity
                * oi.unit_price
                * (1 - oi.discount_percent / 100)
            ),
            0
        ),
        2
    ) AS monetary_value

FROM workspace.default.customers c

LEFT JOIN workspace.default.orders o
    ON c.customer_id = o.customer_id

LEFT JOIN workspace.default.order_items oi
    ON o.order_id = oi.order_id

GROUP BY
    c.customer_id,
    c.customer_name

ORDER BY monetary_value DESC;
```

```python
%sql
WITH customer_rfm AS (
    SELECT
        c.customer_id,
        c.customer_name,

        MAX(o.order_date) AS last_purchase_date,

        COUNT(DISTINCT o.order_id) AS frequency,

        COALESCE(
            SUM(
                oi.quantity
                * oi.unit_price
                * (1 - oi.discount_percent / 100)
            ),
            0
        ) AS monetary_value

    FROM workspace.default.customers c

    LEFT JOIN workspace.default.orders o
        ON c.customer_id = o.customer_id

    LEFT JOIN workspace.default.order_items oi
        ON o.order_id = oi.order_id

    GROUP BY
        c.customer_id,
        c.customer_name
)

SELECT
    customer_id,
    customer_name,
    last_purchase_date,
    frequency,
    ROUND(monetary_value, 2) AS monetary_value,

    CASE
        WHEN frequency = 0 THEN 'No Purchase'
        WHEN frequency = 1 THEN 'One-time Customer'
        WHEN frequency <= 4 THEN 'Occasional Customer'
        ELSE 'Loyal Customer'
    END AS customer_segment

FROM customer_rfm

ORDER BY monetary_value DESC;
```

```python
%sql
WITH customer_metrics AS (
    SELECT
        c.customer_id,
        c.customer_name,

        MAX(o.order_date) AS last_purchase_date,

        COUNT(DISTINCT o.order_id) AS total_orders,

        COALESCE(
            SUM(
                oi.quantity
                * oi.unit_price
                * (1 - oi.discount_percent / 100)
            ),
            0
        ) AS total_spend

    FROM workspace.default.customers c

    LEFT JOIN workspace.default.orders o
        ON c.customer_id = o.customer_id

    LEFT JOIN workspace.default.order_items oi
        ON o.order_id = oi.order_id

    GROUP BY
        c.customer_id,
        c.customer_name
)

SELECT
    customer_id,
    customer_name,
    last_purchase_date,
    total_orders,
    ROUND(total_spend, 2) AS total_spend,

    CASE
        WHEN total_orders = 0 THEN 'No Purchase'
        WHEN total_orders = 1 THEN 'One-time'
        WHEN total_orders <= 4 THEN 'Occasional'
        ELSE 'Loyal'
    END AS purchase_segment,

    CASE
        WHEN total_spend = 0 THEN 'No Purchase'
        WHEN total_spend < 1000 THEN 'Low'
        WHEN total_spend <= 5000 THEN 'Medium'
        ELSE 'High'
    END AS spend_segment

FROM customer_metrics

ORDER BY total_spend DESC;
```

```python
%sql
WITH customer_metrics AS (
    SELECT
        c.customer_id,

        COUNT(DISTINCT o.order_id) AS total_orders,

        COALESCE(
            SUM(
                oi.quantity
                * oi.unit_price
                * (1 - oi.discount_percent / 100)
            ),
            0
        ) AS total_spend

    FROM workspace.default.customers c

    LEFT JOIN workspace.default.orders o
        ON c.customer_id = o.customer_id

    LEFT JOIN workspace.default.order_items oi
        ON o.order_id = oi.order_id

    GROUP BY c.customer_id
)

SELECT
    CASE
        WHEN total_orders = 0 THEN 'No Purchase'
        WHEN total_orders = 1 THEN 'One-time'
        WHEN total_orders <= 4 THEN 'Occasional'
        ELSE 'Loyal'
    END AS customer_segment,

    COUNT(*) AS number_of_customers

FROM customer_metrics

GROUP BY
    CASE
        WHEN total_orders = 0 THEN 'No Purchase'
        WHEN total_orders = 1 THEN 'One-time'
        WHEN total_orders <= 4 THEN 'Occasional'
        ELSE 'Loyal'
    END

ORDER BY number_of_customers DESC;
```
