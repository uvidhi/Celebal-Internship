```python
%sql
SELECT
    c.customer_id,
    c.customer_name,
    ROUND(
        SUM(
            oi.quantity
            * oi.unit_price
            * (1 - oi.discount_percent / 100)
        ),
        2
    ) AS total_revenue
FROM workspace.default.customers c
JOIN workspace.default.orders o
    ON c.customer_id = o.customer_id
JOIN workspace.default.order_items oi
    ON o.order_id = oi.order_id
GROUP BY
    c.customer_id,
    c.customer_name
ORDER BY total_revenue DESC;
```

```python
%sql
SELECT
    c.customer_id,
    c.customer_name,
    COUNT(DISTINCT o.order_id) AS total_orders,
    ROUND(
        SUM(
            oi.quantity
            * oi.unit_price
            * (1 - oi.discount_percent / 100)
        ),
        2
    ) AS total_spend,
    ROUND(
        SUM(
            oi.quantity
            * oi.unit_price
            * (1 - oi.discount_percent / 100)
        )
        / COUNT(DISTINCT o.order_id),
        2
    ) AS average_order_value
FROM workspace.default.customers c
JOIN workspace.default.orders o
    ON c.customer_id = o.customer_id
JOIN workspace.default.order_items oi
    ON o.order_id = oi.order_id
GROUP BY
    c.customer_id,
    c.customer_name
ORDER BY average_order_value DESC;
```

```python
%sql
WITH customer_orders AS (
    SELECT
        c.customer_id,
        c.customer_name,
        COUNT(DISTINCT o.order_id) AS order_count
    FROM workspace.default.customers c
    LEFT JOIN workspace.default.orders o
        ON c.customer_id = o.customer_id
    GROUP BY
        c.customer_id,
        c.customer_name
)

SELECT
    customer_id,
    customer_name,
    order_count,
    CASE
        WHEN order_count = 0 THEN 'No Purchase'
        WHEN order_count = 1 THEN 'One-time'
        WHEN order_count <= 4 THEN 'Occasional'
        ELSE 'Loyal'
    END AS customer_segment
FROM customer_orders
ORDER BY order_count DESC;
```

```python
%sql
WITH customer_spending AS (
    SELECT
        c.customer_id,
        c.customer_name,
        COALESCE(
            SUM(
                oi.quantity
                * oi.unit_price
                * (1 - oi.discount_percent / 100)
            ),
            0
        ) AS total_spend
    FROM workspace.default.customers c
    LEFT JOIN workspace.default.orders o
        ON c.customer_id = o.customer_id
    LEFT JOIN workspace.default.order_items oi
        ON o.order_id = oi.order_id
    GROUP BY
        c.customer_id,
        c.customer_name
)

SELECT
    customer_id,
    customer_name,
    ROUND(total_spend, 2) AS total_spend,

    CASE
        WHEN total_spend = 0 THEN 'No Purchase'
        WHEN total_spend < 1000 THEN 'Low'
        WHEN total_spend <= 5000 THEN 'Medium'
        ELSE 'High'
    END AS spend_tier

FROM customer_spending
ORDER BY total_spend DESC;
```

```python
%sql
WITH customer_metrics AS (
    SELECT
        c.customer_id,
        c.customer_name,

        COUNT(DISTINCT o.order_id) AS order_count,

        COALESCE(
            SUM(
                oi.quantity
                * oi.unit_price
                * (1 - oi.discount_percent / 100)
            ),
            0
        ) AS total_spend

    FROM workspace.default.customers c

    LEFT JOIN workspace.default.orders o
        ON c.customer_id = o.customer_id

    LEFT JOIN workspace.default.order_items oi
        ON o.order_id = oi.order_id

    GROUP BY
        c.customer_id,
        c.customer_name
)

SELECT
    customer_id,
    customer_name,
    order_count,
    ROUND(total_spend, 2) AS total_spend,

    CASE
        WHEN order_count = 0 THEN 'No Purchase'
        WHEN order_count = 1 THEN 'One-time'
        WHEN order_count <= 4 THEN 'Occasional'
        ELSE 'Loyal'
    END AS purchase_segment,

    CASE
        WHEN total_spend = 0 THEN 'No Purchase'
        WHEN total_spend < 1000 THEN 'Low'
        WHEN total_spend <= 5000 THEN 'Medium'
        ELSE 'High'
    END AS spend_segment

FROM customer_metrics
ORDER BY total_spend DESC;
```

```python
%sql
WITH customer_metrics AS (
    SELECT
        c.customer_id,

        COUNT(DISTINCT o.order_id) AS order_count,

        COALESCE(
            SUM(
                oi.quantity
                * oi.unit_price
                * (1 - oi.discount_percent / 100)
            ),
            0
        ) AS total_spend

    FROM workspace.default.customers c

    LEFT JOIN workspace.default.orders o
        ON c.customer_id = o.customer_id

    LEFT JOIN workspace.default.order_items oi
        ON o.order_id = oi.order_id

    GROUP BY c.customer_id
)

SELECT
    CASE
        WHEN order_count = 0 THEN 'No Purchase'
        WHEN order_count = 1 THEN 'One-time'
        WHEN order_count <= 4 THEN 'Occasional'
        ELSE 'Loyal'
    END AS customer_segment,

    COUNT(*) AS number_of_customers

FROM customer_metrics

GROUP BY
    CASE
        WHEN order_count = 0 THEN 'No Purchase'
        WHEN order_count = 1 THEN 'One-time'
        WHEN order_count <= 4 THEN 'Occasional'
        ELSE 'Loyal'
    END

ORDER BY number_of_customers DESC;
```
