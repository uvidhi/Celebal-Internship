```python
query = """
SELECT
    c.customer_id,
    c.customer_name
FROM workspace.default.customers c
LEFT JOIN workspace.default.orders o
    ON c.customer_id = o.customer_id
WHERE o.order_id IS NULL
"""

result = spark.sql(query)

rows = result.collect()

if not rows:
    print("No customers with zero orders found.")
else:
    print("Customers with zero orders:")
    result.show()
```

```python
query = """
SELECT
    COUNT(*) AS customer_count
FROM workspace.default.customers
"""

result = spark.sql(query)

result.show()
```

```python
customer_count = result.collect()[0]["customer_count"]

if customer_count == 1:
    print("The database contains one customer.")
else:
    print("The database contains more than one customer.")
```

```python
query = """
SELECT
    COUNT(*) AS future_orders
FROM workspace.default.orders
WHERE order_date > CURRENT_DATE()
"""

result = spark.sql(query)

future_orders = result.collect()[0]["future_orders"]

print("Future orders:", future_orders)

if future_orders > 0:
    print("Future dates were found and handled.")
else:
    print("No future orders found.")
```
